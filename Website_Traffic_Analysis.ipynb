# Website Traffic Analysis

This notebook analyzes visitor behavior, traffic sources, and website performance using Python, Pandas, Plotly, and Matplotlib.

## Project Overview
- Analyze visitor behavior patterns
- Examine traffic sources
- Evaluate website performance metrics
- Generate insights and visualizations

## Step 4: Import Required Libraries

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import plotly.express as px
from datetime import datetime, timedelta
import warnings
warnings.filterwarnings('ignore')

# Set style for matplotlib
plt.style.use('ggplot')
```

## Step 5: Load the Dataset

```python
# Load your website traffic data
# For local Jupyter: df = pd.read_csv('dataset/website_traffic.csv')
# For Google Colab: df = pd.read_csv('website_traffic.csv')

df = pd.read_csv('dataset/website_traffic.csv')

# Display first few rows
df.head()
```

## Step 6: Check Dataset Information

```python
print("Shape of Dataset:", df.shape)
df.info()
```

## Step 7: Check Missing Values

```python
df.isnull().sum()
```

## Step 8: Statistical Summary

```python
df.describe()
```

## Step 9: Convert Date Column

```python
df["Date"] = pd.to_datetime(df["Date"])
df.head()
```

## 3. Data Cleaning and Preprocessing

```python
# Handle missing values - drop rows with any null values
df = df.dropna()

# Remove duplicates
df = df.drop_duplicates()

# Display cleaned data
print("Data cleaning completed")
print(f"\nCleaned dataset shape: {df.shape}")
print("\nFirst few rows of cleaned data:")
df.head()
```

## 4. Visitor Behavior Analysis

```python
# Analyze visitor behavior metrics
print("=" * 50)
print("VISITOR BEHAVIOR ANALYSIS")
print("=" * 50)

# Total visitors
total_visitors = df['Visitors'].sum()
print(f"\nTotal Visitors: {total_visitors:,.0f}")

# Average session duration
avg_session_duration = df['Session_Duration'].mean()
print(f"Average Session Duration: {avg_session_duration:.2f} minutes")

# Bounce rate
avg_bounce_rate = df['Bounce_Rate'].mean()
print(f"Average Bounce Rate: {avg_bounce_rate:.2f}%")

# Pages per session
avg_pages_per_session = df['Pages_Per_Session'].mean()
print(f"Average Pages Per Session: {avg_pages_per_session:.2f}")

# Daily average visitors
daily_avg_visitors = df['Visitors'].mean()
print(f"Daily Average Visitors: {daily_avg_visitors:.0f}")
```

## 5. Traffic Sources Analysis

```python
# Analyze traffic sources
print("\n" + "=" * 50)
print("TRAFFIC SOURCES ANALYSIS")
print("=" * 50)

# If your data has Traffic_Source column
if 'Traffic_Source' in df.columns:
    traffic_source_summary = df.groupby('Traffic_Source')['Visitors'].sum().sort_values(ascending=False)
    print("\nVisitors by Traffic Source:")
    print(traffic_source_summary)
    
    # Percentage breakdown
    print("\nTraffic Source Breakdown (%):")
    traffic_percentage = (traffic_source_summary / traffic_source_summary.sum() * 100).round(2)
    print(traffic_percentage)
else:
    print("Traffic_Source column not found in dataset")
```

## 6. Website Performance Metrics

```python
# Analyze performance metrics
print("\n" + "=" * 50)
print("WEBSITE PERFORMANCE METRICS")
print("=" * 50)

# Conversion rate
if 'Conversions' in df.columns:
    total_conversions = df['Conversions'].sum()
    conversion_rate = (total_conversions / total_visitors * 100)
    print(f"\nTotal Conversions: {total_conversions:,.0f}")
    print(f"Overall Conversion Rate: {conversion_rate:.2f}%")

# Average time on page
if 'Avg_Time_On_Page' in df.columns:
    avg_time_on_page = df['Avg_Time_On_Page'].mean()
    print(f"Average Time On Page: {avg_time_on_page:.2f} seconds")

# Page load time
if 'Page_Load_Time' in df.columns:
    avg_load_time = df['Page_Load_Time'].mean()
    print(f"Average Page Load Time: {avg_load_time:.2f} seconds")

print(f"\nDate Range: {df['Date'].min().date()} to {df['Date'].max().date()}")
```

## 7. Data Visualizations - Matplotlib

```python
# Create visualizations using Matplotlib
fig, axes = plt.subplots(2, 2, figsize=(15, 10))

# Plot 1: Daily Visitors Over Time
axes[0, 0].plot(df['Date'], df['Visitors'], linewidth=2, color='#2E86AB')
axes[0, 0].set_title('Daily Visitors Over Time', fontsize=12, fontweight='bold')
axes[0, 0].set_xlabel('Date')
axes[0, 0].set_ylabel('Number of Visitors')
axes[0, 0].tick_params(axis='x', rotation=45)
axes[0, 0].grid(True, alpha=0.3)

# Plot 2: Bounce Rate Trend
axes[0, 1].plot(df['Date'], df['Bounce_Rate'], linewidth=2, color='#A23B72')
axes[0, 1].set_title('Bounce Rate Over Time', fontsize=12, fontweight='bold')
axes[0, 1].set_xlabel('Date')
axes[0, 1].set_ylabel('Bounce Rate (%)')
axes[0, 1].tick_params(axis='x', rotation=45)
axes[0, 1].grid(True, alpha=0.3)

# Plot 3: Session Duration Trend
axes[1, 0].plot(df['Date'], df['Session_Duration'], linewidth=2, color='#F18F01')
axes[1, 0].set_title('Average Session Duration Over Time', fontsize=12, fontweight='bold')
axes[1, 0].set_xlabel('Date')
axes[1, 0].set_ylabel('Session Duration (minutes)')
axes[1, 0].tick_params(axis='x', rotation=45)
axes[1, 0].grid(True, alpha=0.3)

# Plot 4: Pages Per Session Trend
axes[1, 1].plot(df['Date'], df['Pages_Per_Session'], linewidth=2, color='#C73E1D')
axes[1, 1].set_title('Pages Per Session Over Time', fontsize=12, fontweight='bold')
axes[1, 1].set_xlabel('Date')
axes[1, 1].set_ylabel('Pages Per Session')
axes[1, 1].tick_params(axis='x', rotation=45)
axes[1, 1].grid(True, alpha=0.3)

plt.tight_layout()
plt.show()

print("Matplotlib visualizations created successfully")
```

## 7. Data Visualizations - Plotly (Interactive)

```python
# Create interactive visualizations using Plotly
# Interactive line chart for Daily Visitors
fig = px.line(df, x='Date', y='Visitors', 
               title='Daily Visitors Over Time (Interactive)',
               labels={'Visitors': 'Number of Visitors', 'Date': 'Date'},
               template='plotly_white')
fig.update_layout(hovermode='x unified', height=500)
fig.show()

# Interactive line chart comparing multiple metrics
fig = px.line(df, x='Date', y=['Bounce_Rate', 'Session_Duration'],
               title='Bounce Rate vs Session Duration',
               labels={'value': 'Value', 'Date': 'Date'},
               template='plotly_white')
fig.update_layout(hovermode='x unified', height=500)
fig.show()

print("Plotly visualizations created successfully")
```

```python
# Traffic Source Pie Chart (if column exists)
if 'Traffic_Source' in df.columns:
    traffic_data = df.groupby('Traffic_Source')['Visitors'].sum().reset_index()
    fig = px.pie(traffic_data, values='Visitors', names='Traffic_Source',
                  title='Website Traffic Distribution by Source',
                  template='plotly_white')
    fig.show()

    # Bar chart for average metrics by traffic source
    source_metrics = df.groupby('Traffic_Source')[['Bounce_Rate', 'Session_Duration', 'Pages_Per_Session']].mean().reset_index()
    fig = px.bar(source_metrics, x='Traffic_Source', y='Bounce_Rate',
                  title='Average Bounce Rate by Traffic Source',
                  template='plotly_white')
    fig.show()
```

## 8. Key Insights and Recommendations

```python
# Document key findings
print("\n" + "=" * 50)
print("KEY INSIGHTS AND RECOMMENDATIONS")
print("=" * 50)

insights = {
    'total_visitors': f"{total_visitors:,.0f}",
    'average_bounce_rate': f"{avg_bounce_rate:.2f}%",
    'average_session_duration': f"{avg_session_duration:.2f} minutes",
    'average_pages_per_session': f"{avg_pages_per_session:.2f}"
}

print("\nKey Metrics:")
for key, value in insights.items():
    formatted_key = key.replace('_', ' ').title()
    print(f"  • {formatted_key}: {value}")

# Generate recommendations based on metrics
print("\nRecommendations:")

if avg_bounce_rate > 50:
    print("  1. High bounce rate detected. Consider improving landing page content and CTA placement.")
else:
    print("  1. Bounce rate is healthy. Continue optimizing user engagement.")

if avg_session_duration < 2:
    print("  2. Session duration is low. Review site navigation and content relevance.")
else:
    print("  2. Good session duration. Users are engaged with your content.")

if avg_pages_per_session < 2:
    print("  3. Users visit few pages per session. Enhance internal linking strategy.")
else:
    print("  3. Pages per session is good. Internal navigation is working well.")

print("  4. Continue monitoring traffic trends and adjust marketing strategy accordingly.")
print("  5. A/B test high-bounce pages to improve conversion rates.")
```
